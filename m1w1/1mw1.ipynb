```python
import tiktoken
enc = tiktoken.get_encoding("cl100k_base")  # GPT-4 tokenizer
text = "The capital of France is Paris."
tokens = enc.encode(text)
print(tokens)           # token IDs
print(enc.decode_tokens_bytes(tokens))  # see actual chunks
print(f"Token count: {len(tokens)}")
```

```python
import os
import numpy as np
from dotenv import load_dotenv
from openai import OpenAI

# Load .env file
load_dotenv()

# Create OpenAI client
client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY")
)

def embed(text):
    response = client.embeddings.create(
        input=text,
        model="text-embedding-3-small"
    )
    return np.array(response.data[0].embedding)

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

king = embed("king")
man = embed("man")
woman = embed("woman")
queen = embed("queen")
print("embed king: ", king)
print("embed king: ", len(king))
print("embed man: ", man)
print("embed woman: ", woman)
print("embed queen: ", queen)

result = king - man + woman

print(
    "king-man+woman vs queen similarity:",
    cosine_similarity(result, queen)
)
7
print(
    "king-man+woman vs paris similarity:",
    cosine_similarity(result, embed("Paris"))
)
```
